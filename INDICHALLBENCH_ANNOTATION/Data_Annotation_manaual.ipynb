```python
import pandas as pd
import numpy as np
```

```python
input_res = pd.read_json('input/responses.json')
input_ref = pd.read_json('input/references.json')
```

```python
output_ann = pd.read_json('output/annotations.jsonl', lines=True)
```

```python
# Check required columns
print("response_id in input_res:", "response_id" in input_res.columns)
print("response_id in output_ann:", "response_id" in output_ann.columns)

# Check duplicate IDs
print("Duplicate response IDs in annotations:",
      output_ann["response_id"].duplicated().sum())

print("Duplicate response records in input:",
      input_res["question_id"].duplicated().sum())
```

```python
import hashlib

def generate_response_id(row):
    raw = "|".join(
        str(row.get(col, ""))
        for col in [
            "question_id",
            "language",
            "model_name",
            "timestamp",
            "response"
        ]
    )
    return hashlib.sha256(raw.encode("utf-8")).hexdigest()[:20]

input_res = input_res.copy()
input_res["response_id"] = input_res.apply(
    generate_response_id, axis=1
)
```

```python
# Normalize IDs before comparing
input_res["response_id"] = (
    input_res["response_id"].astype("string").str.strip()
)

output_ann["response_id"] = (
    output_ann["response_id"].astype("string").str.strip()
)

# Check for missing IDs
print("Missing response IDs in input:", input_res["response_id"].isna().sum())
print("Missing response IDs in annotations:", output_ann["response_id"].isna().sum())

# Build the set of completed annotation IDs
annotated_ids = set(output_ann["response_id"].dropna())

# Mark each response
input_res = input_res.copy()

input_res["annotation_status"] = (
    input_res["response_id"]
    .isin(annotated_ids)
    .map({True: "Annotated", False: "Not Annotated"})
)

# Separate records
annotated_res = input_res[
    input_res["annotation_status"] == "Annotated"
].copy()

unannotated_res = input_res[
    input_res["annotation_status"] == "Not Annotated"
].copy()

print("Total responses:", len(input_res))
print("Matched / annotated:", len(annotated_res))
print("Unannotated:", len(unannotated_res))
print("Total after separation:", len(annotated_res) + len(unannotated_res))
```

```python
print("Unannotated responses by model:")
display(
    unannotated_res["model_name"]
    .value_counts()
    .rename_axis("model_name")
    .reset_index(name="unannotated_count")
)

print("\nUnannotated responses by language:")
display(
    unannotated_res["language"]
    .value_counts(dropna=False)
    .rename_axis("language")
    .reset_index(name="unannotated_count")
)

print("\nModel vs language:")
display(
    pd.crosstab(
        unannotated_res["model_name"],
        unannotated_res["language"],
        dropna=False
    )
)
```

```python
print("Duplicate question IDs in references:",
      input_ref["question_id"].duplicated().sum())
```

```python
manual_review = unannotated_res.merge(
    input_ref,
    on="question_id",
    how="left",
    suffixes=("_response", "_reference"),
    validate="many_to_one",
    indicator=True
)

print("Manual review records:", len(manual_review))

print("\nReference matching status:")
display(manual_review["_merge"].value_counts())

print("\nSample records:")
display(
    manual_review[
        [
            "question_id",
            "language",
            "model_name",
            "question",
            "ground_truth",
            "response"
        ]
    ].head(10)
)
```

```python
# Keep useful columns from both datasets
columns_to_keep = [
    "response_id",
    "question_id",
    "fact_id_response",
    "language",
    "model_name",
    "domain_response",
    "difficulty_response",
    "question",
    "ground_truth",
    "response",
    "source_url_response",
    "source_url_reference",
    "correctness_label",
    "hallucination_types",
    "severity",
    "reason",
    "confidence",
    "review_required",
    "_merge"
]

# Keep only columns that actually exist
columns_to_keep = [
    col for col in columns_to_keep
    if col in manual_review.columns
]

manual_review = manual_review[columns_to_keep].copy()

# Add human annotation fields
manual_review["human_correctness_label"] = ""
manual_review["human_hallucination_types"] = ""
manual_review["human_severity"] = ""
manual_review["human_reason"] = ""
manual_review["human_annotator"] = ""
manual_review["human_review_status"] = "Pending"

display(manual_review.head(10))
```

```python
import os

os.makedirs("output/manual_review", exist_ok=True)

csv_path = "output/manual_review/manual_review_1015.csv"
json_path = "output/manual_review/manual_review_1015.json"

manual_review.to_csv(
    csv_path,
    index=False,
    encoding="utf-8-sig"
)

manual_review.to_json(
    json_path,
    orient="records",
    force_ascii=False,
    indent=2
)

print("Files exported successfully:")
print(csv_path)
print(json_path)
print("Total records exported:", len(manual_review))
```

```python
check_manual = pd.read_json(json_path)

print("Exported rows:", len(check_manual))
print("Unique response IDs:", check_manual["response_id"].nunique())
print("Pending human reviews:",
      (check_manual["human_review_status"] == "Pending").sum())

print("\nMissing questions:",
      check_manual["question"].isna().sum())

print("Missing ground truths:",
      check_manual["ground_truth"].isna().sum())
```
