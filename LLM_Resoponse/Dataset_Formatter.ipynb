```python

import json
from datetime import datetime, timedelta, timezone
from pathlib import Path

# IMPORTANT: Set the exact path of the original JSON file
file_path = Path(
    r"D:\NNDL_Project\Dataset\LLM_Response\responses_zai.json"
).resolve()

IST = timezone(timedelta(hours=5, minutes=30))

print("Target file:", file_path)
print("File exists:", file_path.is_file())

if not file_path.is_file():
    raise FileNotFoundError(f"Original JSON file not found: {file_path}")

# Enter values once
model_name = input("Enter model_name: ").strip()
start_input = input(
    "Enter starting date and time (YYYY-MM-DD HH:MM:SS): "
).strip()

start_time = datetime.strptime(
    start_input, "%Y-%m-%d %H:%M:%S"
).replace(tzinfo=IST)

# Load the original file
with file_path.open("r", encoding="utf-8") as f:
    data = json.load(f)

if not isinstance(data, list):
    raise ValueError("Expected the JSON root to be a list.")

if not all(isinstance(record, dict) for record in data):
    raise ValueError("One or more records are not JSON objects.")

print("Records loaded:", len(data))

# Update every record: +10 minutes after each group of 90
GROUP_SIZE = 90
INTERVAL_MINUTES = 10

for i, record in enumerate(data):
    record["model_name"] = model_name
    offset = (i // GROUP_SIZE) * INTERVAL_MINUTES
    record["timestamp"] = (
        start_time + timedelta(minutes=offset)
    ).isoformat(timespec="seconds")

# Save atomically to the same original file
temp_path = file_path.with_suffix(file_path.suffix + ".tmp")

try:
    with temp_path.open("w", encoding="utf-8", newline="\n") as f:
        json.dump(data, f, ensure_ascii=False, indent=2)
        f.write("\n")
        f.flush()

    # Verify saved JSON before replacing the original
    with temp_path.open("r", encoding="utf-8") as f:
        verified_data = json.load(f)

    assert len(verified_data) == len(data)
    assert verified_data[0]["timestamp"] == start_time.isoformat(
        timespec="seconds"
    )

    if len(verified_data) > GROUP_SIZE:
        assert verified_data[GROUP_SIZE]["timestamp"] == (
            start_time + timedelta(minutes=10)
        ).isoformat(timespec="seconds")

    # Replace the original file
    temp_path.replace(file_path)

except Exception:
    if temp_path.exists():
        temp_path.unlink()
    raise

# Read the original path again to confirm persistence
with file_path.open("r", encoding="utf-8") as f:
    saved_data = json.load(f)

print("\nSUCCESS: Original file updated and verified.")
print("Saved path:", file_path)
print("Total records:", len(saved_data))
print("First timestamp:", saved_data[0]["timestamp"])
print("Record 90:", saved_data[89]["timestamp"])
print("Record 91:", saved_data[90]["timestamp"])
print("Last timestamp:", saved_data[-1]["timestamp"])
print("Model name:", saved_data[0]["model_name"])
```
